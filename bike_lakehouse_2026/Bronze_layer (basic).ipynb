# Building the Bronze Layer
Ingest all raw csv files manually into Delta tables without any kind of transformations.

## load "cust_info.csv"

```python
df = (
    spark.read.option("header","true")
    .option("inferSchema","true")
    .csv("/Volumes/workspace/bronze/raw_sources/source_crm/cust_info.csv")
)
# df.display()
df.write.mode("overwrite").saveAsTable("bronze.crm_cust_info")
```
